### List methods

```python
fruits = ["orange"]

fruits.append("lemon")
# fruits is now: ["orange", "lemon"]

fruits
```

```python
fruits = ["orange", "kiwi"]

fruits.insert(1, "lemon")
# fruits is now: ["orange", "lemon", "kiwi"]

fruits.insert(2, "grapes")

fruits
```
