    training_error.append(1 - training_accuracy)

    y_pred_class = knn.predict(X_test)

# Define a function that accepts a list of features in a DataFrame and returns the RMSE for a test dataset.

#why use a function? so that it is resuable with different features/ x-values

def train_test_rmse(df, feature_cols):  #the 'df' and 'feature_cols' are optional
    X = df[feature_cols]
    y = df.total_rentals

    X_train, X_test, y_train, y_test = train_test_split(X, y,random_state=123)

    linreg = LinearRegression()
    linreg.fit(X_train, y_train)

    y_pred = linreg.predict(X_test)
    return np.sqrt(metrics.mean_squared_error(y_test, y_pred))

# Comparing with Null RMSE
Null RMSE is the RMSE that could be achieved by always predicting the mean response value. It is a benchmark against which you may want to measure your regression model.

Essentially, we are trying to find a good model to help us predict. Null RMSE means just taking the mean value of y - which gives an idea of the 'average' or 'baseline' to compare with if my model cannot beat the null RMSE, then the model is failing to learn any patterns from the features
